```python
# ============================================================
# SECTION 4 — EXPECTED INPUT AND OUTPUT
#
# Required outputs:
# 1. Original | Reconstructed | Reconstruction Error
# 2. Clean | Noisy | Denoised
# 3. VAE Generated Samples
# ============================================================

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import time

from tensorflow.keras import layers, Model
from tensorflow.keras.optimizers import Adam

# ------------------------------------------------------------
# 1. LOAD AND PREPROCESS MNIST
# ------------------------------------------------------------

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Use the laboratory subset
x_train = x_train[:10000]
y_train = y_train[:10000]

x_test = x_test[:2000]
y_test = y_test[:2000]

# Normalize [0,255] -> [0,1]
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Add channel dimension: 28x28 -> 28x28x1
x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)

print("=" * 70)
print("SECTION 4 — EXPECTED INPUT")
print("=" * 70)

print("Training input shape :", x_train.shape)
print("Testing input shape  :", x_test.shape)
print("Single image shape   :", x_train[0].shape)
print("Pixel range          :", x_train.min(), "to", x_train.max())

# Fully connected representation
x_train_flat = x_train.reshape(len(x_train), 784)
x_test_flat = x_test.reshape(len(x_test), 784)

print("FC-AE input shape     :", x_train_flat.shape)


# ============================================================
# PART 1
# ORIGINAL | RECONSTRUCTED | RECONSTRUCTION ERROR
# ============================================================

print("\n" + "=" * 70)
print("PART 1 — ORIGINAL / RECONSTRUCTED / ERROR")
print("=" * 70)

# ------------------------------------------------------------
# Fully Connected Autoencoder
# Architecture:
# 784 -> 128 -> 32 -> 16 -> 32 -> 128 -> 784
# ------------------------------------------------------------

fc_input = layers.Input(shape=(784,))

x = layers.Dense(128, activation="relu")(fc_input)
x = layers.Dense(32, activation="relu")(x)
latent = layers.Dense(16, activation="relu")(x)

x = layers.Dense(32, activation="relu")(latent)
x = layers.Dense(128, activation="relu")(x)

fc_output = layers.Dense(784, activation="sigmoid")(x)

fc_autoencoder = Model(
    fc_input,
    fc_output,
    name="Section4_FC_Autoencoder"
)

fc_autoencoder.compile(
    optimizer=Adam(learning_rate=1e-3),
    loss="binary_crossentropy"
)

# Train
fc_autoencoder.fit(
    x_train_flat,
    x_train_flat,
    epochs=20,
    batch_size=128,
    validation_data=(x_test_flat, x_test_flat),
    shuffle=True,
    verbose=1
)

# Reconstruct test images
fc_reconstructed_flat = fc_autoencoder.predict(
    x_test_flat,
    batch_size=128,
    verbose=0
)

fc_reconstructed = fc_reconstructed_flat.reshape(
    -1, 28, 28, 1
)

# Reconstruction error
fc_error = np.abs(
    x_test - fc_reconstructed
)

# ------------------------------------------------------------
# Display Original | Reconstructed | Error
# ------------------------------------------------------------

fig, axes = plt.subplots(
    3, 10,
    figsize=(16, 5)
)

for i in range(10):

    # Original
    axes[0, i].imshow(
        x_test[i].squeeze(),
        cmap="gray"
    )
    axes[0, i].set_title("Original")
    axes[0, i].axis("off")

    # Reconstruction
    axes[1, i].imshow(
        fc_reconstructed[i].squeeze(),
        cmap="gray"
    )
    axes[1, i].set_title("Reconstructed")
    axes[1, i].axis("off")

    # Reconstruction error
    axes[2, i].imshow(
        fc_error[i].squeeze(),
        cmap="hot"
    )
    axes[2, i].set_title("Error")
    axes[2, i].axis("off")

plt.suptitle(
    "Section 4: Original vs Reconstructed vs Reconstruction Error",
    fontsize=14
)

plt.tight_layout()
plt.show()


# ============================================================
# PART 2
# CLEAN | NOISY | DENOISED
# ============================================================

print("\n" + "=" * 70)
print("PART 2 — CLEAN / NOISY / DENOISED")
print("=" * 70)

# ------------------------------------------------------------
# Add Gaussian noise
# sigma = 0.2
# ------------------------------------------------------------

noise_sigma = 0.2

noise = np.random.normal(
    loc=0.0,
    scale=noise_sigma,
    size=x_train.shape
).astype("float32")

x_train_noisy = np.clip(
    x_train + noise,
    0.0,
    1.0
)

# Test noisy images
test_noise = np.random.normal(
    loc=0.0,
    scale=noise_sigma,
    size=x_test.shape
).astype("float32")

x_test_noisy = np.clip(
    x_test + test_noise,
    0.0,
    1.0
)


# ------------------------------------------------------------
# Denoising Convolutional Autoencoder
# ------------------------------------------------------------

dae_input = layers.Input(
    shape=(28, 28, 1)
)

# Encoder
x = layers.Conv2D(
    32,
    3,
    activation="relu",
    padding="same"
)(dae_input)

x = layers.MaxPooling2D(
    2,
    padding="same"
)(x)

x = layers.Conv2D(
    64,
    3,
    activation="relu",
    padding="same"
)(x)

x = layers.MaxPooling2D(
    2,
    padding="same"
)(x)

x = layers.Conv2D(
    64,
    3,
    activation="relu",
    padding="same"
)(x)

# Decoder
x = layers.UpSampling2D(2)(x)

x = layers.Conv2D(
    32,
    3,
    activation="relu",
    padding="same"
)(x)

x = layers.UpSampling2D(2)(x)

dae_output = layers.Conv2D(
    1,
    3,
    activation="sigmoid",
    padding="same"
)(x)

dae = Model(
    dae_input,
    dae_output,
    name="Section4_Denoising_CAE"
)

dae.compile(
    optimizer=Adam(learning_rate=1e-3),
    loss="binary_crossentropy"
)

# Train noisy input -> clean target
dae.fit(
    x_train_noisy,
    x_train,
    epochs=20,
    batch_size=128,
    validation_data=(x_test_noisy, x_test),
    shuffle=True,
    verbose=1
)

# Denoise test images
x_test_denoised = dae.predict(
    x_test_noisy,
    batch_size=128,
    verbose=0
)


# ------------------------------------------------------------
# Display Clean | Noisy | Denoised
# ------------------------------------------------------------

fig, axes = plt.subplots(
    3, 10,
    figsize=(16, 5)
)

for i in range(10):

    # Clean
    axes[0, i].imshow(
        x_test[i].squeeze(),
        cmap="gray"
    )
    axes[0, i].set_title("Clean")
    axes[0, i].axis("off")

    # Noisy
    axes[1, i].imshow(
        x_test_noisy[i].squeeze(),
        cmap="gray"
    )
    axes[1, i].set_title("Noisy")
    axes[1, i].axis("off")

    # Denoised
    axes[2, i].imshow(
        x_test_denoised[i].squeeze(),
        cmap="gray"
    )
    axes[2, i].set_title("Denoised")
    axes[2, i].axis("off")

plt.suptitle(
    "Section 4: Clean vs Noisy vs Denoised",
    fontsize=14
)

plt.tight_layout()
plt.show()


# ============================================================
# PART 3
# VAE — GENERATED SAMPLES FROM LATENT DISTRIBUTION
# ============================================================

print("\n" + "=" * 70)
print("PART 3 — VAE GENERATED SAMPLES")
print("=" * 70)

latent_dim = 2


# ------------------------------------------------------------
# Sampling layer
# ------------------------------------------------------------

class Sampling(layers.Layer):

    def call(self, inputs):

        z_mean, z_log_var = inputs

        epsilon = tf.random.normal(
            shape=tf.shape(z_mean)
        )

        return (
            z_mean +
            tf.exp(0.5 * z_log_var) *
            epsilon
        )


# ------------------------------------------------------------
# VAE Encoder
# ------------------------------------------------------------

vae_input = layers.Input(
    shape=(28, 28, 1)
)

x = layers.Conv2D(
    32,
    3,
    strides=2,
    activation="relu",
    padding="same"
)(vae_input)

x = layers.Conv2D(
    64,
    3,
    strides=2,
    activation="relu",
    padding="same"
)(x)

x = layers.Flatten()(x)

x = layers.Dense(
    128,
    activation="relu"
)(x)

z_mean = layers.Dense(
    latent_dim,
    name="z_mean"
)(x)

z_log_var = layers.Dense(
    latent_dim,
    name="z_log_var"
)(x)

z = Sampling()([
    z_mean,
    z_log_var
])

vae_encoder = Model(
    vae_input,
    [z_mean, z_log_var, z],
    name="Section4_VAE_Encoder"
)


# ------------------------------------------------------------
# VAE Decoder
# ------------------------------------------------------------

decoder_input = layers.Input(
    shape=(latent_dim,)
)

x = layers.Dense(
    7 * 7 * 64,
    activation="relu"
)(decoder_input)

x = layers.Reshape(
    (7, 7, 64)
)(x)

x = layers.Conv2DTranspose(
    64,
    3,
    strides=2,
    activation="relu",
    padding="same"
)(x)

x = layers.Conv2DTranspose(
    32,
    3,
    strides=2,
    activation="relu",
    padding="same"
)(x)

decoder_output = layers.Conv2D(
    1,
    3,
    activation="sigmoid",
    padding="same"
)(x)

vae_decoder = Model(
    decoder_input,
    decoder_output,
    name="Section4_VAE_Decoder"
)


# ------------------------------------------------------------
# Complete VAE
# ------------------------------------------------------------

class VAE(Model):

    def __init__(
        self,
        encoder,
        decoder,
        **kwargs
    ):
        super().__init__(**kwargs)

        self.encoder = encoder
        self.decoder = decoder

        self.total_loss_tracker = tf.keras.metrics.Mean(
            name="total_loss"
        )

        self.reconstruction_loss_tracker = tf.keras.metrics.Mean(
            name="reconstruction_loss"
        )

        self.kl_loss_tracker = tf.keras.metrics.Mean(
            name="kl_loss"
        )

    @property
    def metrics(self):
        return [
            self.total_loss_tracker,
            self.reconstruction_loss_tracker,
            self.kl_loss_tracker
        ]

    def train_step(self, data):

        if isinstance(data, tuple):
            data = data[0]

        with tf.GradientTape() as tape:

            z_mean, z_log_var, z = self.encoder(
                data,
                training=True
            )

            reconstruction = self.decoder(
                z,
                training=True
            )

            reconstruction_loss = tf.reduce_mean(
                tf.reduce_sum(
                    tf.keras.backend.binary_crossentropy(
                        data,
                        reconstruction
                    ),
                    axis=(1, 2)
                )
            )

            kl_loss = -0.5 * tf.reduce_mean(
                tf.reduce_sum(
                    1 +
                    z_log_var -
                    tf.square(z_mean) -
                    tf.exp(z_log_var),
                    axis=1
                )
            )

            total_loss = (
                reconstruction_loss +
                kl_loss
            )

        gradients = tape.gradient(
            total_loss,
            self.trainable_weights
        )

        self.optimizer.apply_gradients(
            zip(
                gradients,
                self.trainable_weights
            )
        )

        self.total_loss_tracker.update_state(
            total_loss
        )

        self.reconstruction_loss_tracker.update_state(
            reconstruction_loss
        )

        self.kl_loss_tracker.update_state(
            kl_loss
        )

        return {
            "loss": self.total_loss_tracker.result(),
            "reconstruction_loss":
                self.reconstruction_loss_tracker.result(),
            "kl_loss":
                self.kl_loss_tracker.result()
        }


# Create VAE
vae = VAE(
    vae_encoder,
    vae_decoder
)

vae.compile(
    optimizer=Adam(learning_rate=1e-3)
)

# Train
vae.fit(
    x_train,
    epochs=20,
    batch_size=128,
    shuffle=True,
    verbose=1
)


# ------------------------------------------------------------
# Generate new images
# Sample z ~ N(0,I)
# ------------------------------------------------------------

num_generated = 25

random_latent_vectors = np.random.normal(
    size=(num_generated, latent_dim)
).astype("float32")

generated_images = vae_decoder.predict(
    random_latent_vectors,
    verbose=0
)


# ------------------------------------------------------------
# Display VAE generated samples
# ------------------------------------------------------------

fig, axes = plt.subplots(
    5, 5,
    figsize=(8, 8)
)

for i, ax in enumerate(axes.flat):

    ax.imshow(
        generated_images[i].squeeze(),
        cmap="gray"
    )

    ax.set_title(
        f"Sample {i + 1}"
    )

    ax.axis("off")

plt.suptitle(
    "Section 4: VAE Generated Samples from Latent Distribution",
    fontsize=14
)

plt.tight_layout()
plt.show()


# ============================================================
# FINAL SECTION 4 VERIFICATION
# ============================================================

print("\n" + "=" * 70)
print("SECTION 4 — EXPECTED OUTPUT VERIFICATION")
print("=" * 70)

print("1. Original image              : DISPLAYED")
print("2. Reconstructed image         : DISPLAYED")
print("3. Reconstruction error        : DISPLAYED")
print("4. Original clean image        : DISPLAYED")
print("5. Noisy image                 : DISPLAYED")
print("6. Denoised reconstruction     : DISPLAYED")
print("7. VAE generated samples       : DISPLAYED")
print("\nSection 4 completed successfully.")
```

```python
# ============================================================
# SECTION 7 — FULLY CONNECTED AUTOENCODER
# ============================================================

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model

# Load MNIST
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Recommended laboratory subset
x_train = x_train[:10000]
y_train = y_train[:10000]

x_test = x_test[:2000]
y_test = y_test[:2000]

# Normalize pixel values to [0, 1]
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Add channel dimension
x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)

print("Training shape:", x_train.shape)
print("Testing shape :", x_test.shape)

# Flatten images for fully connected AE
x_train_flat = x_train.reshape(x_train.shape[0], 784)
x_test_flat = x_test.reshape(x_test.shape[0], 784)

print("Flattened training shape:", x_train_flat.shape)
print("Flattened testing shape :", x_test_flat.shape)


# ------------------------------------------------------------
# Architecture
# 784 -> 128 -> 32 -> 16 -> 32 -> 128 -> 784
# ------------------------------------------------------------

input_layer = layers.Input(shape=(784,))

x = layers.Dense(
    128,
    activation="relu"
)(input_layer)

x = layers.Dense(
    32,
    activation="relu"
)(x)

latent = layers.Dense(
    16,
    activation="relu",
    name="latent"
)(x)

x = layers.Dense(
    32,
    activation="relu"
)(latent)

x = layers.Dense(
    128,
    activation="relu"
)(x)

output_layer = layers.Dense(
    784,
    activation="sigmoid"
)(x)

autoencoder = Model(
    input_layer,
    output_layer,
    name="Fully_Connected_Autoencoder"
)

autoencoder.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3
    ),
    loss="binary_crossentropy"
)

autoencoder.summary()


# ------------------------------------------------------------
# Train
# ------------------------------------------------------------

import time

fc_start_time = time.time()

history = autoencoder.fit(
    x_train_flat,
    x_train_flat,
    epochs=20,
    batch_size=128,
    validation_data=(x_test_flat, x_test_flat),
    shuffle=True,
    verbose=1
)

fc_training_time = time.time() - fc_start_time

print(
    f"\nFC Autoencoder training time: "
    f"{fc_training_time:.2f} seconds"
)


# ------------------------------------------------------------
# Reconstruction
# ------------------------------------------------------------

x_test_reconstructed_fc = autoencoder.predict(
    x_test_flat,
    verbose=0
)

x_test_reconstructed_fc = x_test_reconstructed_fc.reshape(
    x_test.shape
)

print(
    "Reconstructed shape:",
    x_test_reconstructed_fc.shape
)
```

```python
# ============================================================
# SECTION 9 — RECONSTRUCTION METRICS
# ============================================================

import numpy as np
from skimage.metrics import structural_similarity as ssim

# ------------------------------------------------------------
# MSE
# ------------------------------------------------------------

fc_mse_per_image = np.mean(
    (
        x_test.reshape(x_test.shape[0], -1)
        -
        x_test_reconstructed_fc.reshape(
            x_test_reconstructed_fc.shape[0], -1
        )
    ) ** 2,
    axis=1
)

fc_mse = np.mean(fc_mse_per_image)


# ------------------------------------------------------------
# MAE
# ------------------------------------------------------------

fc_mae_per_image = np.mean(
    np.abs(
        x_test.reshape(x_test.shape[0], -1)
        -
        x_test_reconstructed_fc.reshape(
            x_test_reconstructed_fc.shape[0], -1
        )
    ),
    axis=1
)

fc_mae = np.mean(fc_mae_per_image)


# ------------------------------------------------------------
# SSIM
# ------------------------------------------------------------

fc_ssim_values = []

for i in range(len(x_test)):

    score = ssim(
        x_test[i].squeeze(),
        x_test_reconstructed_fc[i].squeeze(),
        data_range=1.0
    )

    fc_ssim_values.append(score)

fc_ssim = np.mean(fc_ssim_values)


# ------------------------------------------------------------
# Report
# ------------------------------------------------------------

print("===== Fully Connected AE Metrics =====")
print(f"Test MSE : {fc_mse:.6f}")
print(f"Test MAE : {fc_mae:.6f}")
print(f"Mean SSIM: {fc_ssim:.6f}")
```

```python
# ============================================================
# SECTION 10 — CONVOLUTIONAL AUTOENCODER
# ============================================================

from tensorflow.keras import layers, Model

# ------------------------------------------------------------
# Encoder + Decoder
# ------------------------------------------------------------

cae_input = layers.Input(
    shape=(28, 28, 1)
)

x = layers.Conv2D(
    32,
    3,
    activation="relu",
    padding="same"
)(cae_input)

x = layers.MaxPooling2D(
    2,
    padding="same"
)(x)

x = layers.Conv2D(
    64,
    3,
    activation="relu",
    padding="same"
)(x)

x = layers.MaxPooling2D(
    2,
    padding="same"
)(x)

x = layers.Conv2D(
    64,
    3,
    activation="relu",
    padding="same"
)(x)

x = layers.UpSampling2D(
    2
)(x)

x = layers.Conv2D(
    32,
    3,
    activation="relu",
    padding="same"
)(x)

x = layers.UpSampling2D(
    2
)(x)

cae_output = layers.Conv2D(
    1,
    3,
    activation="sigmoid",
    padding="same"
)(x)

cae = Model(
    cae_input,
    cae_output,
    name="Convolutional_Autoencoder"
)

cae.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3
    ),
    loss="binary_crossentropy"
)

cae.summary()


# ------------------------------------------------------------
# Train CAE
# ------------------------------------------------------------

cae_start_time = time.time()

cae_history = cae.fit(
    x_train,
    x_train,
    epochs=20,
    batch_size=128,
    validation_data=(x_test, x_test),
    shuffle=True,
    verbose=1
)

cae_training_time = time.time() - cae_start_time

print(
    f"\nCAE training time: "
    f"{cae_training_time:.2f} seconds"
)


# ------------------------------------------------------------
# Reconstruction
# ------------------------------------------------------------

x_test_reconstructed_cae = cae.predict(
    x_test,
    verbose=0
)

print(
    "CAE reconstructed shape:",
    x_test_reconstructed_cae.shape
)
```

```python
# ============================================================
# SECTION 11 — FC-AE vs CONVOLUTIONAL AE
# ============================================================

import pandas as pd
from skimage.metrics import structural_similarity as ssim

# ------------------------------------------------------------
# Helper function
# ------------------------------------------------------------

def calculate_metrics(original, reconstructed):

    original_flat = original.reshape(
        original.shape[0], -1
    )

    reconstructed_flat = reconstructed.reshape(
        reconstructed.shape[0], -1
    )

    mse = np.mean(
        (original_flat - reconstructed_flat) ** 2
    )

    mae = np.mean(
        np.abs(
            original_flat - reconstructed_flat
        )
    )

    ssim_values = []

    for i in range(len(original)):

        score = ssim(
            original[i].squeeze(),
            reconstructed[i].squeeze(),
            data_range=1.0
        )

        ssim_values.append(score)

    return (
        mse,
        mae,
        np.mean(ssim_values)
    )


# ------------------------------------------------------------
# Metrics
# ------------------------------------------------------------

fc_mse, fc_mae, fc_ssim = calculate_metrics(
    x_test,
    x_test_reconstructed_fc
)

cae_mse, cae_mae, cae_ssim = calculate_metrics(
    x_test,
    x_test_reconstructed_cae
)


# ------------------------------------------------------------
# Parameters
# ------------------------------------------------------------

fc_params = autoencoder.count_params()
cae_params = cae.count_params()


# ------------------------------------------------------------
# Table
# ------------------------------------------------------------

comparison_table = pd.DataFrame({

    "Model": [
        "Fully Connected AE",
        "Convolutional AE"
    ],

    "MSE": [
        fc_mse,
        cae_mse
    ],

    "MAE": [
        fc_mae,
        cae_mae
    ],

    "SSIM": [
        fc_ssim,
        cae_ssim
    ],

    "Parameters": [
        fc_params,
        cae_params
    ]
})

print(
    comparison_table.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# Plot 3
# ------------------------------------------------------------

index = 0

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(
    x_test[index].squeeze(),
    cmap="gray"
)
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(
    x_test_reconstructed_fc[index].squeeze(),
    cmap="gray"
)
plt.title("FC-AE")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(
    x_test_reconstructed_cae[index].squeeze(),
    cmap="gray"
)
plt.title("CAE")
plt.axis("off")

plt.suptitle(
    "Plot 3: Original → FC-AE → CAE"
)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# SECTION 13 — ADDING IMAGE NOISE
# ============================================================

# ------------------------------------------------------------
# Gaussian noise
# ------------------------------------------------------------

def add_gaussian_noise(images, sigma):

    noise = np.random.normal(
        0,
        sigma,
        images.shape
    )

    noisy = images + noise

    return np.clip(
        noisy,
        0,
        1
    ).astype("float32")


# ------------------------------------------------------------
# Salt-and-pepper noise
# ------------------------------------------------------------

def add_salt_pepper_noise(images, probability):

    noisy = images.copy()

    random_values = np.random.random(
        images.shape
    )

    # Salt pixels
    noisy[
        random_values < probability / 2
    ] = 1.0

    # Pepper pixels
    noisy[
        random_values > 1 - probability / 2
    ] = 0.0

    return noisy.astype("float32")


# ------------------------------------------------------------
# Required Gaussian noise levels
# ------------------------------------------------------------

gaussian_sigmas = [0.1, 0.2, 0.3]

gaussian_noisy_train = {}
gaussian_noisy_test = {}

for sigma in gaussian_sigmas:

    gaussian_noisy_train[sigma] = (
        add_gaussian_noise(
            x_train,
            sigma
        )
    )

    gaussian_noisy_test[sigma] = (
        add_gaussian_noise(
            x_test,
            sigma
        )
    )


# ------------------------------------------------------------
# Required salt-and-pepper levels
# ------------------------------------------------------------

sp_levels = [0.05, 0.10, 0.20]

sp_noisy_train = {}
sp_noisy_test = {}

for p in sp_levels:

    sp_noisy_train[p] = (
        add_salt_pepper_noise(
            x_train,
            p
        )
    )

    sp_noisy_test[p] = (
        add_salt_pepper_noise(
            x_test,
            p
        )
    )


print("Gaussian noise levels:", gaussian_sigmas)
print("Salt-and-pepper levels:", sp_levels)
```

```python
# ============================================================
# SECTION 15 — VAE ARCHITECTURE
# ============================================================

import tensorflow as tf
from tensorflow.keras import layers, Model

latent_dim = 2


# ------------------------------------------------------------
# Sampling Layer
# ------------------------------------------------------------

class Sampling(layers.Layer):

    def call(self, inputs):

        z_mean, z_log_var = inputs

        epsilon = tf.random.normal(
            shape=tf.shape(z_mean)
        )

        return (
            z_mean
            +
            tf.exp(0.5 * z_log_var)
            *
            epsilon
        )


# ------------------------------------------------------------
# Encoder
# ------------------------------------------------------------

encoder_input = layers.Input(
    shape=(28, 28, 1),
    name="encoder_input"
)

x = layers.Conv2D(
    32,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(encoder_input)

x = layers.Conv2D(
    64,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(x)

shape_before_flattening = x.shape[1:]

x = layers.Flatten()(x)

x = layers.Dense(
    128,
    activation="relu"
)(x)

z_mean = layers.Dense(
    latent_dim,
    name="z_mean"
)(x)

z_log_var = layers.Dense(
    latent_dim,
    name="z_log_var"
)(x)

z = Sampling()(
    [z_mean, z_log_var]
)

encoder = Model(
    encoder_input,
    [z_mean, z_log_var, z],
    name="encoder"
)


# ------------------------------------------------------------
# Decoder
# ------------------------------------------------------------

decoder_input = layers.Input(
    shape=(latent_dim,),
    name="decoder_input"
)

x = layers.Dense(
    7 * 7 * 64,
    activation="relu"
)(decoder_input)

x = layers.Reshape(
    (7, 7, 64)
)(x)

x = layers.Conv2DTranspose(
    64,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(x)

x = layers.Conv2DTranspose(
    32,
    3,
    strides=2,
    padding="same",
    activation="relu"
)(x)

decoder_output = layers.Conv2D(
    1,
    3,
    padding="same",
    activation="sigmoid"
)(x)

decoder = Model(
    decoder_input,
    decoder_output,
    name="decoder"
)


print("===== ENCODER =====")
encoder.summary()

print("\n===== DECODER =====")
decoder.summary()

print(
    "\nShape before flattening:",
    shape_before_flattening
)
```

```python
# ============================================================
# SECTION 17 — TRAIN THE VAE
# ============================================================

import tensorflow as tf


class VAE(Model):

    def __init__(
        self,
        encoder,
        decoder,
        **kwargs
    ):
        super().__init__(**kwargs)

        self.encoder = encoder
        self.decoder = decoder


    def train_step(self, data):

        if isinstance(data, tuple):
            data = data[0]

        with tf.GradientTape() as tape:

            z_mean, z_log_var, z = self.encoder(
                data,
                training=True
            )

            reconstruction = self.decoder(
                z,
                training=True
            )

            # Reconstruction loss
            bce = tf.keras.backend.binary_crossentropy(
                data,
                reconstruction
            )

            if len(bce.shape) == 4:

                reconstruction_loss = tf.reduce_mean(
                    tf.reduce_sum(
                        bce,
                        axis=(1, 2, 3)
                    )
                )

            else:

                reconstruction_loss = tf.reduce_mean(
                    tf.reduce_sum(
                        bce,
                        axis=(1, 2)
                    )
                )

            # KL loss
            kl_loss = tf.reduce_mean(
                -0.5
                *
                tf.reduce_sum(
                    1
                    + z_log_var
                    - tf.square(z_mean)
                    - tf.exp(z_log_var),
                    axis=1
                )
            )

            total_loss = (
                reconstruction_loss
                +
                kl_loss
            )

        gradients = tape.gradient(
            total_loss,
            self.trainable_weights
        )

        self.optimizer.apply_gradients(
            zip(
                gradients,
                self.trainable_weights
            )
        )

        return {
            "loss": total_loss,
            "reconstruction_loss": reconstruction_loss,
            "kl_loss": kl_loss
        }


    def test_step(self, data):

        if isinstance(data, tuple):
            data = data[0]

        z_mean, z_log_var, z = self.encoder(
            data,
            training=False
        )

        reconstruction = self.decoder(
            z,
            training=False
        )

        bce = tf.keras.backend.binary_crossentropy(
            data,
            reconstruction
        )

        if len(bce.shape) == 4:

            reconstruction_loss = tf.reduce_mean(
                tf.reduce_sum(
                    bce,
                    axis=(1, 2, 3)
                )
            )

        else:

            reconstruction_loss = tf.reduce_mean(
                tf.reduce_sum(
                    bce,
                    axis=(1, 2)
                )
            )

        kl_loss = tf.reduce_mean(
            -0.5
            *
            tf.reduce_sum(
                1
                + z_log_var
                - tf.square(z_mean)
                - tf.exp(z_log_var),
                axis=1
            )
        )

        total_loss = (
            reconstruction_loss
            +
            kl_loss
        )

        return {
            "loss": total_loss,
            "reconstruction_loss": reconstruction_loss,
            "kl_loss": kl_loss
        }


# ------------------------------------------------------------
# Create VAE
# ------------------------------------------------------------

vae = VAE(
    encoder,
    decoder,
    name="Variational_Autoencoder"
)

# IMPORTANT:
# Keras 3 requires a loss when using this custom model setup.
# A dummy loss is supplied because train_step/test_step calculate
# the real VAE losses.

vae.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss=lambda y_true, y_pred:
        tf.reduce_mean(y_pred * 0.0)
)


# ------------------------------------------------------------
# Train
# ------------------------------------------------------------

vae_start_time = time.time()

vae_history = vae.fit(
    x_train,
    epochs=20,
    batch_size=128,
    validation_data=x_test,
    shuffle=True,
    verbose=1
)

vae_training_time = time.time() - vae_start_time

print(
    f"\nVAE training time: "
    f"{vae_training_time:.2f} seconds"
)


# ------------------------------------------------------------
# Final reconstruction
# ------------------------------------------------------------

z_mean_test, z_log_var_test, z_test = encoder.predict(
    x_test,
    batch_size=128,
    verbose=0
)

x_test_reconstructed_vae = decoder.predict(
    z_test,
    batch_size=128,
    verbose=0
)

print(
    "VAE reconstruction shape:",
    x_test_reconstructed_vae.shape
)
```

```python
# ============================================================
# SECTION 27 — ADDITIONAL LATENT-DIMENSION STUDY
# Latent dimensions: {2, 8, 16, 32}
# Report: MSE, SSIM, Parameters
# Plot 10: Latent Dimension vs Reconstruction MSE
# ============================================================

import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras import layers, Model
from tensorflow.keras.optimizers import Adam
from skimage.metrics import structural_similarity as ssim
import time
import gc

# ------------------------------------------------------------
# 1. Check that required data from Section 7 exists
# ------------------------------------------------------------

if "x_train_flat" not in globals() or "x_test_flat" not in globals():
    raise RuntimeError(
        "x_train_flat / x_test_flat not found. "
        "Please run Section 7 first."
    )

print("Training data shape :", x_train_flat.shape)
print("Test data shape     :", x_test_flat.shape)

# ------------------------------------------------------------
# 2. Function to build the basic Fully Connected Autoencoder
# ------------------------------------------------------------

def build_fc_autoencoder(latent_dim):

    inputs = layers.Input(shape=(784,), name="input")

    # Encoder
    x = layers.Dense(128, activation="relu")(inputs)
    x = layers.Dense(32, activation="relu")(x)
    latent = layers.Dense(
        latent_dim,
        activation="relu",
        name=f"latent_{latent_dim}"
    )(x)

    # Decoder
    x = layers.Dense(32, activation="relu")(latent)
    x = layers.Dense(128, activation="relu")(x)
    outputs = layers.Dense(
        784,
        activation="sigmoid"
    )(x)

    model = Model(
        inputs,
        outputs,
        name=f"FC_AE_latent_{latent_dim}"
    )

    model.compile(
        optimizer=Adam(learning_rate=1e-3),
        loss="binary_crossentropy"
    )

    return model


# ------------------------------------------------------------
# 3. Train for each latent dimension
# ------------------------------------------------------------

latent_dimensions = [2, 8, 16, 32]

latent_results = []

models_latent = {}
reconstructions_latent = {}

for latent_dim in latent_dimensions:

    print("\n" + "=" * 60)
    print(f"TRAINING AUTOENCODER — LATENT DIMENSION = {latent_dim}")
    print("=" * 60)

    # Build model
    model = build_fc_autoencoder(latent_dim)

    print("\nModel parameters:",
          f"{model.count_params():,}")

    # Training
    start_time = time.time()

    history = model.fit(
        x_train_flat,
        x_train_flat,
        epochs=20,
        batch_size=128,
        validation_data=(x_test_flat, x_test_flat),
        shuffle=True,
        verbose=1
    )

    training_time = time.time() - start_time

    # Reconstruction
    reconstructed = model.predict(
        x_test_flat,
        batch_size=128,
        verbose=0
    )

    # --------------------------------------------------------
    # MSE
    # --------------------------------------------------------

    mse_value = np.mean(
        (x_test_flat - reconstructed) ** 2
    )

    # --------------------------------------------------------
    # SSIM
    # --------------------------------------------------------

    original_images = x_test_flat.reshape(
        -1, 28, 28
    )

    reconstructed_images = reconstructed.reshape(
        -1, 28, 28
    )

    ssim_values = []

    for i in range(len(original_images)):
        score = ssim(
            original_images[i],
            reconstructed_images[i],
            data_range=1.0
        )
        ssim_values.append(score)

    mean_ssim = np.mean(ssim_values)

    # --------------------------------------------------------
    # Store results
    # --------------------------------------------------------

    latent_results.append({
        "Latent Dimension": latent_dim,
        "MSE": mse_value,
        "SSIM": mean_ssim,
        "Parameters": model.count_params(),
        "Training Time (sec)": training_time
    })

    models_latent[latent_dim] = model
    reconstructions_latent[latent_dim] = reconstructed

    print("\nRESULTS")
    print("-" * 40)
    print(f"Latent Dimension : {latent_dim}")
    print(f"MSE              : {mse_value:.6f}")
    print(f"SSIM             : {mean_ssim:.6f}")
    print(f"Parameters       : {model.count_params():,}")
    print(f"Training Time    : {training_time:.2f} seconds")

    # Free unnecessary TensorFlow memory
    gc.collect()


# ------------------------------------------------------------
# 4. Create final results table
# ------------------------------------------------------------

latent_results_df = pd.DataFrame(latent_results)

print("\n")
print("=" * 70)
print("SECTION 27 — LATENT DIMENSION RESULTS")
print("=" * 70)

print(
    latent_results_df.to_string(
        index=False,
        formatters={
            "MSE": "{:.6f}".format,
            "SSIM": "{:.6f}".format,
            "Training Time (sec)": "{:.2f}".format
        }
    )
)


# ------------------------------------------------------------
# 5. Plot 10 — Latent Dimension vs Reconstruction MSE
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    latent_results_df["Latent Dimension"],
    latent_results_df["MSE"],
    marker="o",
    linewidth=2
)

plt.xlabel("Latent Dimension")
plt.ylabel("Reconstruction MSE")
plt.title("Plot 10: Latent Dimension vs. Reconstruction MSE")

plt.xticks(latent_dimensions)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. Also display SSIM comparison
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    latent_results_df["Latent Dimension"],
    latent_results_df["SSIM"],
    marker="o",
    linewidth=2
)

plt.xlabel("Latent Dimension")
plt.ylabel("Mean SSIM")
plt.title("Latent Dimension vs. Reconstruction SSIM")

plt.xticks(latent_dimensions)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 7. Display example reconstructions
# ------------------------------------------------------------

fig, axes = plt.subplots(
    4, 4,
    figsize=(10, 10)
)

sample_index = 0

for row, latent_dim in enumerate(latent_dimensions):

    original = x_test_flat[sample_index].reshape(28, 28)
    reconstructed = reconstructions_latent[latent_dim][
        sample_index
    ].reshape(28, 28)

    axes[row, 0].imshow(original, cmap="gray")
    axes[row, 0].set_title("Original")
    axes[row, 0].axis("off")

    axes[row, 1].imshow(reconstructed, cmap="gray")
    axes[row, 1].set_title(f"Latent = {latent_dim}")
    axes[row, 1].axis("off")

    # Difference image
    difference = np.abs(
        original - reconstructed
    )

    axes[row, 2].imshow(difference, cmap="hot")
    axes[row, 2].set_title("Absolute Error")
    axes[row, 2].axis("off")

    # Another original/reconstruction pair indicator
    axes[row, 3].text(
        0.5,
        0.5,
        f"Latent = {latent_dim}\n"
        f"MSE = {latent_results_df.loc[row, 'MSE']:.6f}\n"
        f"SSIM = {latent_results_df.loc[row, 'SSIM']:.6f}",
        ha="center",
        va="center",
        fontsize=11
    )
    axes[row, 3].axis("off")

plt.suptitle(
    "Effect of Latent Dimension on Reconstruction",
    fontsize=14
)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 8. Final interpretation helper
# ------------------------------------------------------------

best_mse_row = latent_results_df.loc[
    latent_results_df["MSE"].idxmin()
]

best_ssim_row = latent_results_df.loc[
    latent_results_df["SSIM"].idxmax()
]

print("\n")
print("=" * 70)
print("INTERPRETATION")
print("=" * 70)

print(
    "Lowest reconstruction MSE:"
    f" latent dimension = {int(best_mse_row['Latent Dimension'])},"
    f" MSE = {best_mse_row['MSE']:.6f}"
)

print(
    "Highest SSIM:"
    f" latent dimension = {int(best_ssim_row['Latent Dimension'])},"
    f" SSIM = {best_ssim_row['SSIM']:.6f}"
)

print("""
General observation:
A smaller latent dimension provides stronger compression but may
discard important image information, increasing reconstruction error.
Increasing the latent dimension gives the autoencoder more capacity
to preserve information and generally improves reconstruction quality,
but reduces the degree of compression.
""")
```
